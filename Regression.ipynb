```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

files.upload()
d = pd.read_csv('salary_Data.csv')
X, y = d[['YearsExperience']], d.Salary

# 1. Linear data
m = LinearRegression().fit(X, y)
print("TC1:", m.coef_[0], m.intercept_, "PASS" if m.coef_[0] else "FAIL")

# 2. Train-test
Xt, Xv, yt, yv = train_test_split(X, y, test_size=1/3, random_state=0)
m = LinearRegression().fit(Xt, yt)
p = m.predict(Xv)
print("TC2: MSE =", mean_squared_error(yv,p), "R2 =", r2_score(yv,p),
      "PASS" if r2_score(yv,p)>0.5 else "FAIL")

# 3. Positive correlation
print("TC3:", "PASS" if m.coef_[0]>0 else "FAIL")

# Graphs
plt.scatter(Xt,yt,c='r'); plt.plot(Xt,m.predict(Xt),'b'); plt.show()
plt.scatter(Xv,yv,c='r'); plt.plot(Xv,p,'b'); plt.show()
```
