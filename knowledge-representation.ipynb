```python
from sympy import symbols, Or, Not, Implies, Xor, satisfiable

Rain = symbols("Rain")
Harry_Visited_Hagrid = symbols("Harry_Visited_Hagrid")
Harry_Visited_Dumbledoro = symbols("Harry_Visited_Dumbledoro")

def run_test(testname, kb):
    print("\n" + "=" * 50)
    print(testname)
    print("=" * 50)

    result = satisfiable(kb)
    print("Satisfiable: " + str(result))

    if result is False:
        print("No satisfiable model (contradiction)")
    elif result and Rain in result:
        if result[Rain]:
            print("It rained today")
        else:
            print("There is no rain today")

kb1 = (
    Implies(Not(Rain), Harry_Visited_Hagrid) &
    Xor(Harry_Visited_Hagrid, Harry_Visited_Dumbledoro) &
    Harry_Visited_Dumbledoro
)

run_test("Test Case 1:", kb1)

kb2 = (
    Implies(Not(Rain), Harry_Visited_Hagrid) &
    Xor(Harry_Visited_Hagrid, Harry_Visited_Dumbledoro) &
    Harry_Visited_Hagrid &
    Not(Rain)
)

run_test("Test Case 2:", kb2)

kb3 = (
    Implies(Not(Rain), Harry_Visited_Hagrid) &
    Xor(Harry_Visited_Hagrid, Harry_Visited_Dumbledoro) &
    Harry_Visited_Dumbledoro &
    Rain
)
run_test("TEST CASE 3", kb3)

kb4 = (
    Implies(Not(Rain), Harry_Visited_Hagrid) &
    Xor(Harry_Visited_Hagrid, Harry_Visited_Dumbledoro) &
    Harry_Visited_Hagrid&
    Harry_Visited_Dumbledoro
)
run_test("TEST CASE 4", kb4)

kb5 = Implies(Not(Rain), Harry_Visited_Hagrid)
run_test("TEST CASE 5", kb5)
```
