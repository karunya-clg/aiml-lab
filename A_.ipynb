```python

def AStarAlgo(start_node, stop_node, Graph_nodes, H_dist):
    open_set = {start_node}
    closed_set = set()

    g = {start_node: 0}
    parents = {start_node: start_node}

    while open_set:
        # Find node with minimum f(n) = g(n) + h(n)
        n = min(open_set, key=lambda v: g[v] + H_dist[v])

        # Goal reached
        if n == stop_node:
            path = []
            while parents[n] != n:
                path.append(n)
                n = parents[n]

            path.append(start_node)
            path.reverse()

            print("Path found:", path)
            print("Total cost:", g[stop_node])
            return path, g[stop_node]

        # Explore neighbours
        for m, weight in Graph_nodes.get(n, []):
            tentative_g = g[n] + weight

            if tentative_g < g.get(m, float('inf')):
                g[m] = tentative_g
                parents[m] = n

                if m in closed_set:
                    closed_set.remove(m)

                open_set.add(m)

        open_set.remove(n)
        closed_set.add(n)

    print("Path not found")
    return None, float("inf")


# Test Case 1
Graph_nodes = {
    'S': [('A', 1), ('B', 4)],
    'A': [('G', 1)],
    'B': [('G', 4)],
    'G': []
}

H_dist = {
    'S': 0,
    'A': 100,
    'B': 1,
    'G': 0
}

AStarAlgo('S', 'G', Graph_nodes, H_dist)


# Test Case 2
Graph_nodes = {
    'S': [('A', 2), ('B', 4)],
    'A': [('G', 5)],
    'B': [],
    'G': []
}

H_dist = {
    'S': 3,
    'A': 2,
    'B': 1,
    'G': 0
}

AStarAlgo('S', 'G', Graph_nodes, H_dist)


# Test Case 3
Graph_nodes = {
    'S': [('G', 3)],
    'G': []
}

H_dist = {
    'S': 1,
    'G': 0
}

AStarAlgo('S', 'G', Graph_nodes, H_dist)
```
