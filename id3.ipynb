```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.preprocessing import LabelEncoder


def run(data, name):
    print("\n-----------------------------")
    print(name)
    print("-----------------------------")

    # Convert data into DataFrame
    df = pd.DataFrame(data)

    # Separate input features and target
    X = df.iloc[:, :-1].copy()
    y = df.iloc[:, -1].copy()

    # Convert categorical features into numbers
    for col in X.columns:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))

    # Convert target into numbers
    le = LabelEncoder()
    y = le.fit_transform(y.astype(str))

    # Create Decision Tree using Entropy
    model = DecisionTreeClassifier(
        criterion="entropy",
        random_state=42
    )

    # Train the model
    model.fit(X, y)

    # Display tree rules
    print("Decision Tree Rules:")
    print(export_text(model, feature_names=list(X.columns)))

    # Display depth and leaves
    print("Tree Depth:", model.get_depth())
    print("Number of Leaves:", model.get_n_leaves())


# Test Case 1: Play Tennis
play_tennis = [
    ["Sunny", "Hot", "High", "No"],
    ["Sunny", "Hot", "High", "No"],
    ["Overcast", "Hot", "High", "Yes"],
    ["Rain", "Mild", "High", "Yes"],
    ["Rain", "Cool", "Normal", "Yes"],
    ["Rain", "Cool", "Normal", "No"],
    ["Overcast", "Cool", "Normal", "Yes"],
    ["Sunny", "Mild", "High", "No"]
]

run(play_tennis, "Test Case 1: Play Tennis")


# Test Case 2: Clear Class Separation
clear_data = [
    [1, "Low", "No"],
    [2, "Low", "No"],
    [8, "High", "Yes"],
    [9, "High", "Yes"]
]

run(clear_data, "Test Case 2: Clear Class Separation")


# Test Case 3: Mixed Attributes
mixed_data = [
    ["Red", 10, "Yes"],
    ["Blue", 20, "No"],
    ["Red", 15, "Yes"],
    ["Green", 25, "No"],
    ["Blue", 12, "Yes"]
]

run(mixed_data, "Test Case 3: Mixed Attributes")


# Test Case 4: Same Class
same_class = [
    ["A", 10, "Yes"],
    ["B", 20, "Yes"],
    ["C", 30, "Yes"],
    ["D", 40, "Yes"]
]

run(same_class, "Test Case 4: Same Class")


# Test Case 5: Missing Attribute
missing_data = [
    ["Red", 10, "Yes"],
    ["Blue", None, "No"],
    ["Green", 20, "Yes"]
]

try:
    run(missing_data, "Test Case 5: Missing Attribute")
except Exception as e:
    print("\nError in Missing Attribute Test Case:")
    print(e)
```
