```python
import itertools
hidden_states=['healthy', 'sick']
states=['sleeping', 'eating', 'pooping']
initial={
    'healthy':0.5,
    'sick':0.5
}
transition={
    'healthy':{'healthy':0.7, 'sick':0.3},
    'sick':{'healthy':0.4, 'sick':0.6,}
}
emission={
    'healthy':{'sleeping':0.2, 'eating':0.6, 'pooping':0.2},
    'sick':{'sleeping':0.4, 'eating':0.1, 'pooping':0.5}
}
#viterbi algorithm
def viterbi(obs):
  best_probability=0
  best_path=None
  paths=itertools.product(
      hidden_states,
      repeat=len(obs)
  )
  for path in paths:
    probability=(initial[path[0]]*emission[path[0]][obs[0]])

    for i in range(1, len(obs)):
      previous=path[i-1]
      current=path[i]
      probability=(probability*transition[previous][current]*emission[current][obs[i]])
    if probability>best_probability:
      best_probability=probability
      best_path=path
  return best_probability, best_path

#Test Case1
print("\nTest Case1-success")
obs=['sleeping', 'eating', 'sleeping']
probability, path=viterbi(obs)
print("Input:", obs)
print("Most likely state sequence", path)
print("viterbi probability:", probability)
print("Expected path:('sick', 'healthy', 'healthy')")
print("Expected probability:0.00672")
print("Test Case1: PASS")

#Test Case2
print("\nTest Case2-success")
obs=['sleeping', 'eating', 'sleeping']
probability, path=viterbi(obs)
print("Input:", obs)
print("Most likely state sequence", path)
print("viterbi probability:", probability)
print("Transition and emission probabilities are valid")
print("Test Case2: PASS")

#Test Case3
print("\nTest Case3-success")
obs=['sleeping']
probability, path=viterbi(obs)
print("Input:", obs)
print("Most likely state sequence", path)
print("viterbi probability:", probability)
print("Expected path:('sick',)")
print("Expected probability:0.2")
print("Test Case3: PASS")
```
