```python
def graph_colouring(states, neighbors, colors):
    colour = {}

    def promising(state, col):
        for neighbor in neighbors.get(state, []):
            if neighbor in colour and colour[neighbor] == col:
                return False
        return True

    def get_colour(state):
        for col in colors:
            if promising(state, col):
                return col
        return None

    for state in states:
        col = get_colour(state)
        if col is None:
            return("Coloring not possible")
        colour[state] = col

    return colour

print("TEST CASE 1")
states = ['a', 'b', 'c', 'd']
neighbors = {
    'a': ['b', 'c','d'],
    'b': ['a', 'd'],
    'c': ['a', 'd'],
    'd': ['b', 'c', 'a']
}
colors = ['red', 'green', 'blue']
result = graph_colouring(states, neighbors, colors)
print(result)

print("TEST CASE 2")
states1 = ['a', 'b', 'c']
neighbors1 = {
    'a': ['b', 'c'],
    'b': ['a', 'c'],
    'c': ['a', 'b']
}
colors1 = ['red', 'green', 'blue']
result = graph_colouring(states1, neighbors1, colors1)
print(result)

print("TEST CASE 3")
states2 = ['a', 'b', 'c', 'd']
neighbors2 = {
    'a': ['c', 'd'],
    'b': ['c', 'd'],
    'c': ['a', 'b'],
    'd': ['a', 'b']
}
colors2 = ['red', 'green']
result = graph_colouring(states2, neighbors2, colors2)
print(result)

print("TEST CASE 4")
states3 = ['a', 'b', 'c', 'd']
neighbors3 = {
    'a': ['b', 'c', 'd'],
    'b': ['a', 'c', 'd'],
    'c': ['a', 'b', 'd'],
    'd': ['a', 'b', 'c']
}
colors3 = ['red', 'green', 'blue']
result = graph_colouring(states3, neighbors3, colors3)
print(result)
```
