```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np
import matplotlib.pyplot as plt

iris_dataset = load_iris()
X, y = iris_dataset.data, iris_dataset.target
# TC1: LINEARLY SEPARABLE DATA
X1, y1 = X[y != 2], y[y != 2]

model1 = SVC(kernel='linear')
model1.fit(X1, y1)

print("TC1 Accuracy:", accuracy_score(y1, model1.predict(X1)))

# TC2: TWO-CLASS DATASET
X_train, X_test, y_train, y_test = train_test_split(
    X1, y1, test_size=0.2, random_state=42
)
model2 = SVC(kernel='linear')
model2.fit(X_train, y_train)

print("TC2 Accuracy:", accuracy_score(y_test, model2.predict(X_test)))
#test point near margin
model3 = SVC(kernel='linear')
model3.fit(X1, y1)
point=model3.support_vectors_[0].reshape(1,-1)
print("TC3 Predicted Class:", model3.predict(point)[0])
print("TC3 Decision Value:", model3.decision_function(point)[0])

# Graph
Xg = X1[:, [0, 1]]
mg = SVC(kernel='linear').fit(Xg, y1)

plt.scatter(Xg[y1 == 0, 0], Xg[y1 == 0, 1], label="Class 0")
plt.scatter(Xg[y1 == 1, 0], Xg[y1 == 1, 1], label="Class 1")

xx = np.linspace(Xg[:, 0].min(), Xg[:, 0].max(), 100)
w, b = mg.coef_[0], mg.intercept_[0]

plt.plot(xx, -(w[0]*xx+b)/w[1], label="Hyperplane")
plt.plot(xx, -(w[0]*xx+b-1)/w[1], '--', label="Margin +1")
plt.plot(xx, -(w[0]*xx+b+1)/w[1], '--', label="Margin -1")

plt.scatter(mg.support_vectors_[:, 0], mg.support_vectors_[:, 1],
            facecolors='none', edgecolors='black', s=100,
            label="Support Vectors")

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("SVM Classification")
plt.legend()
plt.grid()
plt.show()
```
