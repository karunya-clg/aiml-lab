```python
from itertools import permutations

def tsp(graph, start):
    n = len(graph)

    if n == 1:
        return 0, [start]
    vertices = []

    for i in range(n):
        if i != start:
            vertices.append(i)

    min_cost = float('inf')
    best_path = []

    for p in permutations(vertices):
        current_cost = 0
        current_city = start
        current_path = [start]

        for city in p:
            current_cost += graph[current_city][city]
            current_city = city
            current_path.append(city)

        current_cost += graph[current_city][start]
        current_path.append(start)

        if current_cost < min_cost:
            min_cost = current_cost
            best_path = current_path

    return min_cost, best_path

print("TEST CASE 1")
graph = [
    [0]
]
start = 0
cost, path = tsp(graph, start)
print("Minimum Tour Cost =", cost)
print("Shortest Path =", path)

print("TEST CASE 2")
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]
start = 0
cost, path = tsp(graph, start)
print("Minimum Tour Cost =", cost)
print("Shortest Path =", path)
```
